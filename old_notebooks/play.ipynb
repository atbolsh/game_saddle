# Interactive play (mode 1)

Have a live conversation with the agent over **one persistent multi-gold
room**. Same scenario controls as `multi_gold_eval.ipynb` (gold count,
openings, New room, End game), **no analyst**. One **Ask** is one
generation: the move is applied immediately. `[END_GAME]` freezes the
session (`session_state` leaves `active`; Ask greys out until **New
room**). Eating gold does not end the session. After a move, the
next Ask prefixes the player question with a Board update (ate /
did not eat) and stores that text on the user message, same as
the self-eval player path.

The **architecture** / **checkpoint** dropdowns switch models the same
way as the other notebooks (`MODEL_CANDIDATES.md`,
`training/TRAINING_OVERVIEW.md`). Switching is an explicit **Switch
model** press.

The live row is the current frame on the left, teal settings stacked
over the maroon scratchpad on the right. Settings stay a solid teal
summary until Edit. Settings: the full scene (pose,
gold, walls, radii, gameSize), a direction slider (0…2π), and
add/remove rows for gold, openings (side + center + width), and
walls; from/to are computed
on Render. Scratchpad is still JSON — bad format stays in
edit mode. Generation cannot start while either card is being
edited. Scene
and scratchpad edits are hidden from the agent until the next Ask
(the next generation always snapshots the live board and notepad).
The `-- starting board --` print under the row is scrollback from
cell start; Render updates the live current-board widget, not that
history image.

**Takeover** writes the player reply yourself. It stays on until
**Resume agent**.

Prereqs:
- Neo4j is up (`bash scripts/vast_neo4j_launch.sh`, or `docker compose up -d neo4j`).
- The semantic model is seeded once (`python -m agent.runner seed`).
- `bash scripts/setup_env.sh` (installs deps incl. `ipywidgets`).

The first cell connects NAMS and loads Gemma, so it can take a minute.


```python
import os
import sys

# Run from the repo root so `agent` imports and `.env` resolve.
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
sys.path.insert(0, os.getcwd())

from agent.multi_gold_session import MultiGoldPlaySession

# Connects NAMS on a background loop and loads the default model (first run is slow).
# Logging is on by default: every LLM generate call and every DB retrieval for
# this session is written under session.logger.run_dir (see the "Dump DB
# status" button in the panel below).
session = MultiGoldPlaySession(n_gold=None, opening="require")
print("session_id:", session.session_id)
print("session_state:", session.session_state)
if session.logger is not None:
    print("log dir:", session.logger.run_dir)
```

```python
import html as _html
import json
import re

import ipywidgets as widgets
from IPython.display import Image, clear_output, display

from agent import game_io
from agent.notebook_ui import (
    UiBusy, live_board_row, model_picker, player_takeover_controls,
    room_scenario_bar, tame_shift_enter,
)
from agent.modes import DEFAULT_PLAYER_QUESTION

_TARGET_RE = re.compile(r"^TARGET:\s*(.+)$", re.MULTILINE)

busy = UiBusy()
room = room_scenario_bar()
gold_dd, opening_dd = room.gold_dd, room.opening_dd
new_room_btn, end_btn = room.new_room_btn, room.end_btn

question_box = widgets.Textarea(
    value=DEFAULT_PLAYER_QUESTION,
    placeholder="Ask the player about the current scene...",
    description="You:",
    layout=widgets.Layout(width="600px", height="70px"),
)
ask_btn = widgets.Button(description="Ask", button_style="primary")
restart_btn = widgets.Button(description="Restart conversation", button_style="warning")
dump_btn = widgets.Button(description="Dump DB status", button_style="info")
out = widgets.Output(layout=widgets.Layout(width="100%"))
banner = widgets.HTML()
FRAME_WIDTH = 420

_last_player_raw = None


def _current_target(raw):
    if not raw:
        return "(none)"
    matches = _TARGET_RE.findall(raw)
    return matches[-1].strip() if matches else "(none)"


def _openings_summary():
    d = game_io.game_to_settings_dict(session.game)
    ops = d.get("openings") or []
    if not ops:
        return "none (sealed)"
    bits = [
        f"{o['side']} w={o['width']:.3f} @ ({o['center'][0]:.2f},{o['center'][1]:.2f})"
        for o in ops
    ]
    return "; ".join(bits)


def _refresh_banner(last_raw=None):
    golds = game_io.gold_remaining(session.game)
    banner.value = (
        "<div style='padding:6px 10px;border:1px solid #888;background:#f7f7f7;"
        "font-family:monospace'>"
        f"<b>state</b>={session.session_state}"
        f" &nbsp; <b>golds</b>={golds}"
        f" &nbsp; <b>openings</b>={_html.escape(_openings_summary())}"
        f" &nbsp; <b>TARGET</b>={_html.escape(_current_target(last_raw))}"
        "</div>"
    )


takeover = None
live = None


def _sync_phase():
    active = session.session_state == "active"
    editing = live is not None and live.is_editing()
    gen = busy.generating
    player_on = active and not gen and not editing
    question_box.disabled = not player_on
    ask_btn.disabled = not player_on or (takeover is not None and takeover.is_on())
    end_btn.disabled = (not active) or gen
    new_room_btn.disabled = gen
    restart_btn.disabled = gen
    dump_btn.disabled = gen
    if live is not None:
        live.sync_buttons()
    if takeover is not None:
        takeover.sync(player_on)


busy.set_on_change(_sync_phase)
live = live_board_row(session, busy, width=FRAME_WIDTH, on_changed=_sync_phase)


def _show_frame(path, caption):
    if path:
        print(caption)
        display(Image(filename=path, width=FRAME_WIDTH))


def _show_step(step):
    i = step["step"]
    if i == 0:
        _show_frame(step["before_path"], "-- frame the agent saw --")
    for s in step.get("searches", []):
        print(f"  [SEARCH {s['query']}]")
    print(f"Agent: {step['raw']}")
    if step["action"] == "END_GAME":
        print("[END_GAME -- session ended; press New room to continue]")
    elif step["action"]:
        print(
            f"[move {i}: {step['action']}  "
            f"gold_collected={step['gold_collected']}  "
            f"gold_remaining={step['gold_remaining']}]"
        )
        _show_frame(step["after_path"], f"-- board after move {i} ({step['action']}) --")
    elif step.get("bare_move"):
        print(
            f"[FORMAT ERROR: bare '{step['bare_move']}' without brackets -- "
            f"NOT a move; turn ends]"
        )
    else:
        print("[no move: answer / end of turn]")


def _apply_room_dropdowns():
    session.n_gold = gold_dd.value
    session.opening = opening_dd.value


def _ask(q, human_reply=None):
    global _last_player_raw
    print(f"\n=== You: {q}")
    result = session.ask(q, on_step=_show_step, human_reply=human_reply)
    step = result["steps"][0]
    _last_player_raw = step["raw"]
    print(f"\n[gold_remaining={result['gold_remaining']}]")
    if result.get("session_state") and result["session_state"] != "active":
        print(f">>> session_state={result['session_state']}; press New room to continue.")
    _refresh_banner(_last_player_raw)
    live.refresh()
    question_box.value = DEFAULT_PLAYER_QUESTION


def on_ask(_):
    q = question_box.value.strip()
    if not q:
        return
    if live.is_editing():
        return
    busy.set_generating(True)
    try:
        with out:
            _ask(q)
    finally:
        busy.set_generating(False)


def on_human_submit(raw):
    q = question_box.value.strip() or DEFAULT_PLAYER_QUESTION
    if live.is_editing():
        return
    busy.set_generating(True)
    takeover.sync(False)
    try:
        with out:
            _ask(q, human_reply=raw)
    finally:
        busy.set_generating(False)


takeover = player_takeover_controls(on_human_submit, on_mode_change=_sync_phase)


def on_new_room(_):
    global _last_player_raw
    _apply_room_dropdowns()
    busy.set_generating(True)
    try:
        with out:
            info = session.reset_game()
            _last_player_raw = None
            print("\n*** New room. ***")
            print(f"n_gold={session.n_gold!r}  opening={session.opening!r}")
            print("openings oracle:")
            print(json.dumps(session.current_settings_dict().get("openings"), indent=2))
            _show_frame(info["frame_path"], "-- board after New room --")
            _refresh_banner()
        live.refresh(force_views=True)
    finally:
        busy.set_generating(False)


def on_end(_):
    busy.set_generating(True)
    try:
        with out:
            info = session.force_end("user")
            print("\n*** Session ended by user. ***")
            _show_frame(info["frame_path"], "-- board at end --")
            _refresh_banner(_last_player_raw)
            print(">>> Press New room to continue.")
        live.refresh()
    finally:
        busy.set_generating(False)


def on_restart(_):
    global _last_player_raw
    _apply_room_dropdowns()
    info = session.restart()
    with out:
        clear_output()
        print(f"New conversation. session_id={info['session_id']}")
        print("openings oracle:")
        print(json.dumps(session.current_settings_dict().get("openings"), indent=2))
        _show_frame(info["frame_path"], "-- fresh board --")
        _refresh_banner()
    question_box.value = DEFAULT_PLAYER_QUESTION
    _last_player_raw = None
    live.refresh(force_views=True)
    _sync_phase()


def on_dump(_):
    dump_btn.disabled = True
    try:
        with out:
            info = session.dump_db()
            print(f"\nDB dumped -> {info['path']}")
            print(f"  nodes = {info['nodes']}   relationships = {info['relationships']}")
            if session.logger is not None:
                print(f"  llm log: {session.logger.llm_txt}")
                print(f"  db  log: {session.logger.db_txt}")
    finally:
        dump_btn.disabled = False


ask_btn.on_click(on_ask)
new_room_btn.on_click(on_new_room)
end_btn.on_click(on_end)
restart_btn.on_click(on_restart)
dump_btn.on_click(on_dump)


def _on_model_switched(info):
    global _last_player_raw
    with out:
        if info["restarted"]:
            clear_output()
            print(f"New conversation under {info['label']}. "
                  f"session_id={session.session_id}")
            _show_frame(session.current_frame_path(), "-- fresh board --")
            _refresh_banner()
            live.refresh(force_views=True)
            _last_player_raw = None
            question_box.value = DEFAULT_PLAYER_QUESTION
        else:
            print(f"\n[model switched to {info['label']} -- conversation continues]")
    _sync_phase()


display(widgets.VBox([
    model_picker(session, on_switched=_on_model_switched),
    room.box,
    banner,
    live.box,
    out,
    question_box,
    widgets.HBox([ask_btn, takeover.takeover_btn, takeover.resume_btn,
                  restart_btn, dump_btn]),
    takeover.controls_box,
]))
tame_shift_enter(
    question_box, takeover.reply_box,
    live.scratch.textarea,
)

_sync_phase()
_refresh_banner()
with out:
    print(f"session_id={session.session_id}")
    print("openings oracle:")
    print(json.dumps(session.current_settings_dict().get("openings"), indent=2))
    _show_frame(session.current_frame_path(), "-- starting board --")
```

### Restore the DB to "semantic seeding only" (optional cleanup)

A conversation writes episodic nodes to Neo4j (`Conversation`, `Message`,
`GameSnapshot`, `ReasoningTrace`/`ReasoningStep`). If a run went badly and you
want the *status quo ante* — a fresh box right after `python -m agent.runner
seed` + `link` — run the cell below. It deletes **all** of those episodic nodes
and keeps only the seeded semantic model (`Entity` + `Preference` nodes and
their relationships). After the wipe it re-heals the current `core_player_*` /
`core_analyst_*` prompt tips from the code seed. Optionally clears the on-disk
snapshot images, and starts a fresh conversation.

The cell is gated by `if False:` so it never runs by accident — flip it to
`if True:` (or run the body manually) when you actually want to reset. Note it
clears **every** conversation, not just the most recent one.

```python
# Flip to `if True:` to wipe episodic memory and restore the seeded state.
if False:
    import shutil

    # 1. Delete every non-semantic node; keep Entity + Preference (+ their edges).
    deleted = session.reset_memory_to_seed()
    print("deleted nodes:", deleted)

    # 2. (Optional) drop the on-disk snapshot PNGs -- all orphaned after step 1.
    img_dir = session.cfg.image_dir
    if img_dir.exists():
        shutil.rmtree(img_dir, ignore_errors=True)
        print("cleared image dir:", img_dir)

    # 3. Start a fresh conversation thread + board so you can keep playing.
    info = session.restart()
    print("new session_id:", info["session_id"])
```

When you're done, release the model and close the memory client:
